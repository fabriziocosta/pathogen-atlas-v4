# Inspect the two-dimensional profile space

This notebook reads the published profile coordinates and renders a lightweight SVG scatter plot without coupling the plot to a particular provider or embedding implementation.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
while ROOT != ROOT.parent and not (ROOT / 'pathogen_atlas_v4').is_dir():
    ROOT = ROOT.parent
if not (ROOT / 'pathogen_atlas_v4').is_dir():
    raise RuntimeError('Open this notebook from the V4 repository')
sys.path.insert(0, str(ROOT))
print(ROOT)
```

```python
import gzip, json
from html import escape
payload = json.loads(gzip.open(ROOT / 'sites/pathogen-atlas-explorer/explorer-data.json.gz', 'rt', encoding='utf-8').read())
profiles = [p for p in payload['species'] if p.get('organism_group') == 'bacteria' and p.get('x') is not None and p.get('y') is not None]
len(profiles), payload['groups']
```

```python
width, height, pad = 900, 560, 36
xs = [float(p['x']) for p in profiles]; ys = [float(p['y']) for p in profiles]
x0,x1=min(xs),max(xs); y0,y1=min(ys),max(ys)
points=[]
for p in profiles:
    x=pad+(float(p['x'])-x0)/(x1-x0)*(width-2*pad)
    y=height-pad-(float(p['y'])-y0)/(y1-y0)*(height-2*pad)
    points.append(f'<circle cx="{x:.1f}" cy="{y:.1f}" r="3" fill="#20877d" fill-opacity=".58"><title>{escape(p["accepted_name"])}</title></circle>')
svg=f'<svg xmlns="http://www.w3.org/2000/svg" width="{width}" height="{height}" viewBox="0 0 {width} {height}"><rect width="100%" height="100%" fill="white"/>{"".join(points)}<text x="{width/2}" y="{height-8}" text-anchor="middle" font-family="sans-serif">Embedding 1</text><text x="18" y="{height/2}" transform="rotate(-90 18 {height/2})" text-anchor="middle" font-family="sans-serif">Embedding 2</text></svg>'
from IPython.display import SVG, display
display(SVG(svg))
```

V4 native embeddings should be stored with a versioned feature profile and run ID. This notebook currently inspects the released human baseline coordinates; it does not claim those coordinates were generated by the V4 embedding pipeline.
